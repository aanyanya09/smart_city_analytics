```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df_traf = pd.read_csv('traffic.csv')
print(df_traf.head())
df_air = pd.read_csv('air_quality_dataset.csv')
print(df_air.head())
```

```python

# Display Top 5 Rows of Traffic Dataset
df_traf.head()

# Display Top 5 Rows of Air Quality Dataset
df_air.head()
```

```python
df_traf.info()
df_air.info()
```

```python
df_traf.describe()
df_air.describe()
```

```python
print(df_traf.isnull().sum())

print(df_air.isnull().sum())
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df_traf['DateTime'][:200],
    df_traf['Vehicles'][:200]
)

plt.title('Traffic Volume Over Time')
plt.xlabel('Time')
plt.ylabel('Vehicle Count')

plt.show()
```

```python
plt.figure(figsize=(12,7))

# Average Traffic per Junction
df_traf.groupby('Junction')['Vehicles'].mean().plot(
    kind='bar',
    width=0.7
)

# Titles and Labels
plt.title(
    'Average Traffic Density per Junction',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel(
    'Junction',
    fontsize=13
)

plt.ylabel(
    'Average Number of Vehicles',
    fontsize=13
)

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Rotate x-axis labels
plt.xticks(rotation=0)

# Layout
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,7))

# Correlation Matrix
corr_matrix = df_air[
    ['PM2.5', 'PM10', 'NO2',
     'Temperature', 'Humidity']
].corr()

# Heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    linewidths=1
)

# Title
plt.title(
    'Correlation Heatmap of Air Quality Factors',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,7))

# Histogram
plt.hist(
    df_air['PM2.5'],
    bins=25,
    color='skyblue',
    edgecolor='black',
    linewidth=1.2,
    alpha=0.9
)

# Title
plt.title(
    'Distribution of PM2.5 Pollution Levels',
    fontsize=18,
    fontweight='bold',
    pad=15
)

# Axis Labels
plt.xlabel(
    'PM2.5 Levels',
    fontsize=14,
    fontweight='bold'
)

plt.ylabel(
    'Frequency',
    fontsize=14,
    fontweight='bold'
)

# Grid
plt.grid(
    linestyle='--',
    alpha=0.5
)

# Background Color
plt.gca().set_facecolor('#f8f9fa')

# Tick Styling
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

# Layout
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14,7))

# Customized Boxplot
sns.boxplot(
    x='Junction',
    y='Vehicles',
    data=df_traf,
    palette='Set2',
    linewidth=2,
    width=0.6
)

# Title
plt.title(
    'Traffic Volatility and Outliers per Junction',
    fontsize=18,
    fontweight='bold',
    pad=15
)

# Axis Labels
plt.xlabel(
    'Junction',
    fontsize=14,
    fontweight='bold'
)

plt.ylabel(
    'Number of Vehicles',
    fontsize=14,
    fontweight='bold'
)

# Grid
plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.5
)

# Background Color
plt.gca().set_facecolor('#f8f9fa')

# Tick Styling
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

# Layout
plt.tight_layout()

plt.show()
```
